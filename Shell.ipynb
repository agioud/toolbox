# Commande Terminal

```python
📂 Navigation
pwd              → dossier actuel
ls / ls -l / ls -a → liste fichiers
cd dossier       → entrer dossier
cd ..            → niveau au-dessus
cd ~             → home
cd -             → dossier précédent

📁 Fichiers & Dossiers
touch file            → créer fichier
code file             → créer fichier vs
mkdir dir             → créer dossier
mkdir -p a/b/c        → créer arborescence

cp file copy          → copier fichier
cp -r dir copy        → copier dossier

mv file dir/          → déplacer
mv old new            → renommer

rm file               → supprimer fichier
rm -r dir             → supprimer dossier
rm -rf dir            → suppression forcée ⚠️

📖 Lecture fichiers
cat file        → afficher contenu
less file       → lecture scrollable
head file       → début fichier
tail file       → fin fichier
tail -f log     → suivi temps réel

🔎 Recherche
find . -name "file"       → chercher fichier
grep "text" file          → chercher texte
grep -r "text" folder     → recherche récursive

⚙️ Permissions
chmod +x script.sh    → rendre exécutable
chmod 755 file        → modifier droits
chown user file       → changer propriétaire

📦 Processus
ps           → liste processus
top / htop   → monitoring live
kill PID     → stop processus
kill -9 PID  → stop forcé

🌐 Réseau
ping site.com      → tester connexion
curl URL           → requête HTTP
wget URL           → télécharger fichier

💾 Système & stockage
df -h          → espace disque
du -sh dir     → taille dossier
free -h        → RAM
uname -a       → infos système

🔄 Redirections & Pipes ⭐
cmd > file        → sortie vers fichier (écrase)
cmd >> file       → ajoute sortie
cmd1 | cmd2       → pipeline


Exemple :

ls -l | grep ".txt"

🧠 Variables
echo $VAR
export VAR=value
env

⌨️ Raccourcis Terminal
Ctrl + C    → stop
Ctrl + D    → quitter
Ctrl + L    → clear
Ctrl + R    → recherche historique
Tab         → autocomplétion

🔥 Commandes très utiles
sudo cmd     → admin
man cmd      → documentation
which cmd    → chemin commande
history      → historique
alias ll="ls -l"
!!           → dernière commande

⭐ Rappels importants
⚠️ rm -rf peut supprimer TOUT
💡 utiliser TAB souvent
💡 Ctrl + R = énorme gain de temps
```

# Commande Git

| Commande Git                | Utilité                                                     |
| --------------------------- | ----------------------------------------------------------- |
| `git init`                  | Initialise un nouveau dépôt Git dans le dossier courant     |
| `git clone <url>`           | Clone un dépôt distant (GitHub, GitLab, etc.) en local      |
| `git status`                | Affiche l’état des fichiers (modifiés, ajoutés, non suivis) |
| `git add <fichier>`         | Ajoute un fichier à l’index (zone de staging)               |
| `git add .`                 | Ajoute tous les fichiers modifiés à l’index                 |
| `git commit -m "message"`   | Enregistre les changements avec un message                  |
| `git log`                   | Affiche l’historique des commits                            |
| `git log --oneline`         | Historique des commits en version courte                    |
| `git diff`                  | Montre les différences entre fichiers modifiés              |
| `git branch`                | Liste les branches existantes                               |
| `git branch <nom>`          | Crée une nouvelle branche                                   |
| `git checkout <branche>`    | Change de branche                                           |
| `git checkout -b <branche>` | Crée et change de branche en une seule commande             |
| `git switch <branche>`      | Alternative moderne à `checkout` pour changer de branche    |
| `git merge <branche>`       | Fusionne une branche dans la branche courante               |
| `git pull`                  | Récupère et fusionne les changements du dépôt distant       |
| `git push`                  | Envoie les commits locaux vers le dépôt distant             |
| `git remote -v`             | Affiche les dépôts distants configurés                      |
| `git reset <fichier>`       | Retire un fichier de l’index                                |
| `git reset --hard`          | Annule tous les changements locaux (⚠️ destructif)          |
| `git stash`                 | Met de côté les modifications non commit                    |
| `git stash pop`             | Restaure les modifications mises de côté                    |
| `git fetch`                 | Récupère les changements distants sans fusion               |
| `git rm <fichier>`          | Supprime un fichier et l’enregistre dans Git                |
| `git tag`                   | Crée ou liste des tags (versions)                           |


```python
# Ajout d'une branche local au depot distant

git switch -b nouvelle-branche   # créer la branche
git add .
git commit -m "mon commit"
git push --set-upstream origin nouvelle-branche
```

# Commande conda 


| Commande                               | Description                                                  |
| -------------------------------------- | ------------------------------------------------------------ |
| `conda create -n env_name`             | Créer un nouvel environnement                                |
| `conda create -n env_name python=3.10` | Créer un environnement avec une version spécifique de Python |
| `conda activate env_name`              | Activer un environnement                                     |
| `conda deactivate`                     | Désactiver l’environnement actuel                            |
| `conda env list`                       | Lister tous les environnements                               |
| `conda info --envs`                    | Voir les environnements (alternative)                        |
| `conda remove -n env_name --all`       | Supprimer un environnement                                   |
| `conda install package_name`           | Installer un package                                         |
| `conda install package_name=1.2.3`     | Installer une version spécifique                             |
| `conda remove package_name`            | Supprimer un package                                         |
| `conda list`                           | Lister les packages installés                                |
| `conda update conda`                   | Mettre à jour Conda                                          |
| `conda update --all`                   | Mettre à jour tous les packages                              |
| `conda search package_name`            | Rechercher un package                                        |
| `conda env export > environment.yml`   | Exporter un environnement                                    |
| `conda env create -f environment.yml`  | Recréer un environnement depuis un fichier                   |
